**7. Basic Statistical Operations**

```python
print("Statistical summary:")
print(df.describe())
```

For example, calculate the mean of numerical columns:

```python
print("Mean:")
print(df.mean(numeric_only=True))
```

Calculate the minimum:

```python
print("Minimum:")
print(df.min(numeric_only=True))
```

Calculate the maximum:

```python
print("Maximum:")
print(df.max(numeric_only=True))
```

**10. Filtering Data**

For example, display flowers having petal length greater than 5:

```python
result = df[df["PetalLengthCm"] > 5]

print(result)
```

Multiple conditions can also be used:

```python
result = df[
    (df["PetalLengthCm"] > 5) &
    (df["SepalLengthCm"] > 6)
]

print(result)
```

**11. Sorting Data**

Sort according to petal length:

```python
sorted_df = df.sort_values("PetalLengthCm")

print(sorted_df)
```

Sort from highest to lowest:

```python
sorted_df = df.sort_values(
    "PetalLengthCm",
    ascending=False
)

print(sorted_df)
```

**12. Counting Different Species**

```python
print(df["Species"].value_counts())
```

This tells us how many records belong to each species.

**13. NumPy Operation**

Convert the petal-length column into a NumPy array:

```python
petal_length = np.array(df["PetalLengthCm"])

print("Petal length:")
print(petal_length)
```

Calculate the mean using NumPy:

```python
print("Mean petal length:", np.mean(petal_length))
```

Calculate the standard deviation:

```python
print("Standard deviation:", np.std(petal_length))
```
